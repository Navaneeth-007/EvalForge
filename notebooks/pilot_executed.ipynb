```python
import json
repo = Path('/kaggle/working/EvalForge')
smoke = json.loads((repo / 'outputs/eval_results.json').read_text())
print('Smoke check completed on', smoke['n'], 'examples; these are debugging results, not a benchmark.')
for path in [repo / 'data/processed', repo / 'outputs']:
    path.rename(path.with_name(path.name + '_smoke'))
subprocess.run(['git', 'pull', '--ff-only'], cwd=str(repo), check=True)
pilot = subprocess.run(['/kaggle/temp/evalforge-env/bin/python', '-u', '-m', 'scripts.run_gpu', '--sample-size', '1000', '--epochs', '1', '--eval-n', '200'], cwd=str(repo), env=os.environ.copy())
print('Pilot experiment exit code:', pilot.returncode)
```

```python
import json, shutil, hashlib, subprocess, os
from pathlib import Path
assert pilot.returncode == 0, 'Pilot failed; inspect the preceding output before packaging'
repo = Path('/kaggle/working/EvalForge')
subprocess.run(['git', 'pull', '--ff-only'], cwd=str(repo), check=True)
python = '/kaggle/temp/evalforge-env/bin/python'
tables = next((repo / 'data/spider_archive').rglob('tables.json'))
subprocess.run([python, '-m', 'eval.rescore', '--report', 'outputs/eval_results.json', '--db-dir', str(tables.parent / 'database'), '--output', 'outputs/eval_results_verified.json'], cwd=str(repo), env=os.environ.copy(), check=True)
subprocess.run([python, '-m', 'scripts.publish_evidence', '--manifest', 'outputs/phi3-sql-qlora/training_manifest.json', '--report', 'outputs/eval_results_verified.json'], cwd=str(repo), env=os.environ.copy(), check=True)
staging = Path('/kaggle/working/evalforge-evidence')
staging.mkdir(exist_ok=True)
shutil.copytree(repo / 'outputs/phi3-sql-qlora/final', staging / 'adapter', dirs_exist_ok=True)
for source in [repo / 'outputs/phi3-sql-qlora/training_manifest.json', repo / 'outputs/eval_results.json', repo / 'outputs/eval_results_verified.json', repo / 'data/processed/eval_samples.json', repo / 'app/evidence/pilot.json']:
    shutil.copy2(source, staging / source.name)
archive = shutil.make_archive(str(staging), 'zip', staging)
evidence = json.loads((staging / 'pilot.json').read_text())
print('PILOT_EXPERIMENT:', json.dumps(evidence['experiment']))
print('PILOT_BASE_METRICS:', json.dumps(evidence['metrics']['base']))
print('PILOT_ADAPTER_METRICS:', json.dumps(evidence['metrics']['adapter']))
print('EVIDENCE_ARCHIVE:', archive)
print('ARCHIVE_SHA256:', hashlib.sha256(Path(archive).read_bytes()).hexdigest())
```
